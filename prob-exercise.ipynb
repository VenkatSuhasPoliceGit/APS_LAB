```python
import random


def coin_experiment(n):
    toss = [random.choice(['head', 'tail']) for _ in range(n)]

    head = toss.count('head')
    tail = toss.count('tail')

    p_head = head / n
    p_tail = tail / n

    print("heads:", head)
    print("tails:", tail)
    print("experimental probability of head:", p_head)
    print("experiemental probability of tail:", p_tail)

coin_experiment(100)

```

```python

for i in [10, 100, 1000, 10000]:
    coin_experiment(i)
```

```python


def count_die(n):
    rolls=[random.randint(1,6) for _ in range(n)]

    for i in range(1,7):
        count=rolls.count(i)
        prob=count/n

        print(f"{i} has count {count} with probability {prob}")


count_die(600)
```

```python
def check_addition_rule(n):
    rolls=[random.randint(1,6) for _ in range(n)]

    count_A=rolls.count(1)
    count_B=rolls.count(6)

    total_count=count_A+count_B

    prob_A=count_A/n
    prob_B=count_B/n

    total_prob=total_count/n

    print(f"Total trials: {n}")
    print(f"Count(A = 1): {count_A}")
    print(f"Count(B = 1): {count_B}")

    print(f"P(A) = {prob_A}")
    print(f"P(B) = {prob_B}")
    print(f"P(A ∪ B) = {total_prob}")

    print(f"P(A) + P(B) = {(prob_A + prob_B)}")

    if abs(total_prob - (prob_A + prob_B)) < 0.0001:
        print(" A and B are disjoint events.")
    else:
        print("Not disjoint.")

check_addition_rule(1000)
```

```python
def probability(count, total):
    return count/total

s={1,2,3,4,5,6}

a={2,4,6}
b={4,5,6}

a_union_b=a|b
print("a U b: ",a_union_b)

a_intersection_b=a&b
print("a ∩ b: ", a_intersection_b)

prob_a=probability(len(a), len(s))
prob_b=probability(len(b), len(s))

print("prob of a: ", prob_a)
print("prob of b: ", prob_b)

p_union=probability(len(a_union_b), len(s))
p_intersection=probability(len(a_intersection_b), len(s))


print("prob of a U b:", p_union)
print("prob of a ∩ b:", p_intersection)

print("p(a)+p(b)-p(a∩b):", prob_a+prob_b-p_intersection)

```

```python

def experiment(n):
    heads = 0
    even = 0
    both = 0

    for _ in range(n):
        toss = random.choice(["Head", "Tail"])
        roll = random.randint(1, 6)

        if toss == "Head":
            heads += 1


        if roll % 2 == 0:
            even += 1

        if toss == "Head" and roll % 2 == 0:
            both += 1

    return heads, even, both



n = 10000

heads, even, both = experiment(n)

P_A = heads / n
P_B = even / n
P_A_intersection_B = both / n

print("P(A) =", P_A)
print("P(B) =", P_B)
print("P(A ∩ B) =", P_A_intersection_B)

print("P(A) × P(B) =", P_A * P_B)
```
